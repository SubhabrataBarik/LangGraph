Running this check first isolates environment/installation problems from logic problems:

- If this notebook fails, the issue is with the environment (installation, kernel selection, virtualenv activation) and must be fixed before anything else is attempted.
- If this notebook succeeds, any later failures can be attributed to the workflow code itself rather than to missing dependencies.

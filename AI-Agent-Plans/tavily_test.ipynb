```python
from langchain_community.tools.tavily_search import TavilySearchResults
from dotenv import load_dotenv
import os
# Notebook lives in AI-Agent-Plans/, but the .env is in the project root.
# Resolve the project root relative to this notebook so the key is found
# no matter which subfolder the kernel was started from.
_PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
load_dotenv(os.path.join(_PROJECT_ROOT, ".env"))
assert os.getenv("TAVILY_API_KEY"), "TAVILY_API_KEY missing from .env"
```
